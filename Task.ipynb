```python
# 1. function that prints "Hello World"
def hello_world():
    return"Hello, World!"

print(hello_world())
```

```python
# 2. function that greet with name
def greet(name):
    return f"Hello, {name}!"

print(greet("Jubal"))
```

```python
# 3. function for square of a number
def square(num):
    return num ** 2

print(square(5))
```

```python
# 4. Sum of two numbers
def add(a, b):
    return a + b

print(add(3, 7))
```

```python
# 5. function for maximum of two numbers
def maximum(a, b):
    return max(a, b)

print(maximum(10, 4))
```

```python
# 6. function for even or odd
def even_or_odd(num):
    return "Even" if num % 2 == 0 else "Odd"

print(even_or_odd(9))
```

```python
# 7. function length of a string
def string_length(text):
    return len(text)

print(string_length("President"))
```

```python
# 8. function for converting Celsius to Fahrenheit
def celsius_to_fahrenheit(c):
    return (c * 9/5) + 32

print(celsius_to_fahrenheit(0))
```

```python
# 9. function for Average of a list
def average(numbers):
    return sum(numbers) / len(numbers)

print(average([10, 20, 30]))
```

```python
# 10. functions to Count vowels
def count_vowels(text):
    vowels = "aeiouAEIOU"
    return sum(1 for char in text if char in vowels)

print(count_vowels("Education"))
```

```python
# 11. function for Unique values in a list
def unique_list(lst):
    return list(set(lst))

print(unique_list([1, 2, 2, 3, 4, 4]))
```

```python
# 12. function for Palindrome check
def is_palindrome(word):
    return word == word[::-1]

print(is_palindrome("civic"))
```

```python
# 13. function for Simple interest
def simple_interest(p, r, t):
    return (p * r * t) / 100

print(simple_interest(1000, 5, 2))
```

```python
# 14. function for Tax calculation with default rate
def calculate_tax(income):
    if income <= 150000:
        tax = income * 0.10
    else:
        tax = income * 0.20
    return tax

income = 200000
print(calculate_tax(income))
```

```python
# 15. function for Factorial
def factorial(n):
    result = 1
    for i in range(1, n + 1):
        result *= i
    return result

print(factorial(5))
```

```python
# 16. Function for Sum of any number of arguments
def sum_all(*args):
    return sum(args)

print(sum_all(1, 2, 3, 4, 5, 6))
```

```python
# 17. function for Minimum value in a list
def minimum_value(lst):
    return min(lst)

print(minimum_value([5, 3, 8, 1]))
```

```python
# 18. function that converts minutes to hours and minutes 
def convert_minutes(minutes):
    hours = minutes // 60
    remaining_minutes = minutes % 60
    return hours, remaining_minutes

h,m = convert_minutes(160)
print(h, "hours", m, "minutes")
```

```python
# 20. function for Area of a rectangle
def rectangle_area(length, width):
    return length * width

print(rectangle_area(5, 4))
```

```python
# 21. function for Largest number in a list
def largest_number(lst):
    return max(lst)

print(largest_number([2, 9, 1, 7]))
```

```python
# 22. function for Positive, negative, or zero
def number_type(num):
    if num > 0:
        return "Positive"
    elif num < 0:
        return "Negative"
    else:
        return "Zero"

print(number_type(-3))
```

```python
# 23. function to Reverse a string
def reverse_string(text):
    return text[::-1]

print(reverse_string("King"))
```

```python
# 24. function for Power of a number
def power(base, exponent):
    return base ** exponent

print(power(2, 3))
```

```python
# 25. function for Grade calculation
def grade(score):
    if score >= 70:
        return "A"
    elif score >= 60:
        return "B"
    elif score >= 50:
        return "C"
    elif score >= 45:
        return "D"
    else:
        return "F"

print(grade(75))
```

```python
# 26. function to Count words in a sentence
def count_words(sentence):
    return len(sentence.split())

print(count_words("Python is very easy"))
```

```python
# 27. function for Prime number check
def is_prime(num):
    if num <= 1:
        return False
    for i in range(2, int(num ** 0.5) + 1):
        if num % i == 0:
            return False
    return True

print(is_prime(11))
```

```python
# 28. function for Sum of digits
def sum_of_digits(num):
    return sum(int(d) for d in str(abs(num)))

print(sum_of_digits(1234))
```

```python
# 29. function to Apply discount
def apply_discount(prices, discount=0.1):
    return [price * (1 - discount) for price in prices]

print(apply_discount([100, 200, 300]))
```

```python
# 30. function for Full name formatter
def full_name(first, last):
    return f"{first} {last}"

print(full_name("Jubal", "Sambo"))
```
